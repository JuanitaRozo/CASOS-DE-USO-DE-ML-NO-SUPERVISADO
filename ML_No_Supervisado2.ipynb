```python
compras = [
    {"pan", "leche", "huevos"}, {"pan", "leche"},
    {"pan", "mantequilla"}, {"leche", "huevos"},
    {"pan", "leche", "mantequilla"}, {"pan", "huevos"},
    {"leche", "mantequilla"}, {"pan", "leche", "huevos"}
]
n = len(compras)
n_pan = sum("pan" in c for c in compras)
n_pan_leche = sum({"pan", "leche"}.issubset(c) for c in compras)
n_leche = sum("leche" in c for c in compras)

soporte = n_pan_leche / n
confianza = n_pan_leche / n_pan
soporte_leche = n_leche / n
lift = confianza / soporte_leche

print(f"Total de compras: {n}")
print(f"Soporte de pan y leche: {soporte:.2f}")
print(f"Confianza pan -> leche: {confianza:.2f}")
print(f"Lift pan -> leche: {lift:.2f}")
```
