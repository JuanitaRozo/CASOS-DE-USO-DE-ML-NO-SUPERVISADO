```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

iris = load_iris()
X = StandardScaler().fit_transform(iris.data)
pca = PCA(n_components=2)
X_2d = pca.fit_transform(X)

print("Forma original:", iris.data.shape)
print("Forma reducida:", X_2d.shape)
print("Varianza explicada por componente:", pca.explained_variance_ratio_)
print("Varianza total explicada:", pca.explained_variance_ratio_.sum())
plt.figure(figsize=(8,5))
for clase, nombre in enumerate(iris.target_names):
    plt.scatter(X_2d[iris.target == clase, 0],
                X_2d[iris.target == clase, 1],
                label=nombre, s=45)
plt.xlabel("Componente principal 1")
plt.ylabel("Componente principal 2")
plt.title("PCA: conjunto Iris en dos dimensiones")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
